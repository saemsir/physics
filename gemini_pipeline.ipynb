```python
# 8. Video Play Karein
import glob
from IPython.display import Video, display

video_files = glob.glob('output/colab-render/**/*.mp4', recursive=True)

if video_files:
    print(f"✅ Found {len(video_files)} video(s). Playing:")
    display(Video(video_files[0], embed=True, width=800))
else:
    print("❌ No videos found. Please check the logs in Cell 7.")
```
